# Netflix Movies & TV Shows — Exploratory Data Analysis
**CodeAlpha Data Analytics Internship — Task 2**

This notebook explores the Netflix titles dataset (8,807 entries) to uncover patterns in content type, country distribution, ratings, duration, and trends over time.

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("netflix_titles.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset =['date_added'])
```

```python
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
```

```python
df['duration'] = df['duration'].fillna(df['duration'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip())
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
```

```python
df.head()
```

```python
df['duration_int'] = df['duration'].str.extract(r'(\d+)').astype(int)
df['duration_type'] = df['duration'].apply(lambda x: 'Season' if 'Season' in x else 'Minute')
```

```python
df.head()
```

```python
df.to_csv("netflix_cleaned.csv", index=False)
```

```python
df[['duration', 'duration_int', 'duration_type', 'date_added', 'year_added', 'month_added']].head(10)
```

# Exploratory Data Analysis — Questions & Insights

Now that the dataset is cleaned, let's explore it by asking meaningful questions about 
content type, country distribution, ratings, duration, and trends over time.

## Q1: How many Movies vs TV Shows are there?

```python
df['type'].value_counts()
```

## Q2: What are the oldest and newest titles on Netflix?

```python
df[['title','release_year']].sort_values('release_year').head(2)
```

```python
df[['title','release_year']].sort_values('release_year').tail(2)
```

## Q3: Which month do most titles get added to Netflix?

```python
df['month_added'].value_counts().sort_index()
```

## Q4: What's the single longest movie on Netflix?

```python
df[df['duration_type']=='Minute'].sort_values('duration_int', ascending=False)[['title','duration']].head(3)
```

## Q5: Which TV show has the most seasons?

```python
df[df['duration_type']=='Season'].sort_values('duration_int', ascending=False)[['title','duration']].head(3)
```

## Q6: What's the longest title name on Netflix?

```python
df['title_length'] = df['title'].apply(len)
df.sort_values('title_length', ascending=False)[['title']].head(3)
```

## Q7: What percentage of total content is Movies vs TV Shows?

```python
df['type'].value_counts(normalize=True) * 100
```

## Q8. Which countries have the most content?

```python
df['country'].value_counts().head(3)
```

## Q9: What's the most common content rating overall?

```python
df['rating'].value_counts().head(3)
```

## Q10: How many titles were added in the most recent year (2021) compared to the very first year in the data?

```python
df['year_added'].value_counts().sort_index()[[df['year_added'].min(), df['year_added'].max()]]
```

# Key Takeaways
- Movies dominate the catalog (~70%) though TV Shows have grown steadily
- Content skews mature (TV-MA is the top rating)
- The US and India lead in content volume, with ~9% of entries missing country data
- December and July are peak months for new content additions
